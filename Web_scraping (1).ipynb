```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import time
```

```python
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (Chrome/120.0.0.0 Safari/537.36)'}


web_page=requests.get('https://www.goodreads.com/shelf/show/fiction?page=1',headers=headers)
```

```python
soup=BeautifulSoup(web_page.text,'lxml')
```

```python
Books=soup.find_all('div',class_='elementList')
```

```python
name=[]
rating=[]
author=[]

for i in Books:
    title=i.find('a',class_='bookTitle')
    stars=i.find('span',class_='greyText smallText')
    writer=i.find('span',itemprop='name')
    
    name.append(title.text.strip() if title else None)
    rating.append(stars.text.strip() if stars else None)
    author.append(writer.text.strip() if writer else None)
    
```

```python
d={'name':name,'rating':rating,'author':author}
```

```python
df=pd.DataFrame(d)
```

```python
df.head()
```

```python
all_data = []

for j in range(1, 501):
    url = f'https://www.goodreads.com/shelf/show/fiction?page={j}'
    web_page = requests.get(url, headers=headers)
    soup = BeautifulSoup(web_page.text, 'lxml')
    Books = soup.find_all('div', 'elementList')
    
    for i in Books:
        title  = i.find('a', class_='bookTitle')
        stars  = i.find('span', class_='greyText smallText')
        writer = i.find('span', itemprop='name')
        
        all_data.append({
            'name':   title.text.strip() if title else None,
            'rating': stars.text.strip() if stars else None,
            'author': writer.text.strip() if writer else None,
        })
    
    time.sleep(0.5)
    print(f'Page {j} done')

df = pd.DataFrame(all_data)
print(df.shape)
```

```python
df = pd.DataFrame(all_data)
```

```python
df['rating'] = df['rating'].astype(str)
df['rating'] = df['rating'].str.replace('\n', '').str.strip()
df['avg_rating'] = df['rating'].str.extract(r'avg rating (\d+\.\d+)')
df['num_ratings'] = df['rating'].str.extract(r'—\s*([\d,]+)\s*ratings')
df['year_published'] = df['rating'].str.extract(r'published (\d+)')
```

```python
df['avg_rating'] = df['avg_rating'].astype(float)
df['year_published'] = df['year_published'].fillna(0).astype(int)
df.dtypes
```

```python
df.head()
```

```python
def get_rating_category(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    elif rating >= 3.5:
        return 'Average'
    else:
        return 'Below Average'

df['rating_category'] = df['avg_rating'].apply(get_rating_category)
```

```python
def get_era(year):
    if year < 1900:
        return 'Classic'
    elif year < 1970:
        return 'Modern'
    elif year < 2000:
        return 'Contemporary'
    else:
        return 'Recent'

df['era'] = df['year_published'].apply(get_era)
```

```python
def get_popularity(num):
    if num >= 1000000:
        return 'Viral'
    elif num >= 500000:
        return 'Very Popular'
    elif num >= 100000:
        return 'Popular'
    else:
        return 'Niche'

df['popularity'] = df['num_ratings'].apply(get_popularity)
```

```python
df.sample(5)
```

```python
df['clean_title'] = df['name'].str.replace(r'\(.*?\)', '', regex=True).str.strip()
```

```python
def get_century(year):
    if year < 1800:
        return '18th century or earlier'
    elif year < 1900:
        return '19th century'
    elif year < 2000:
        return '20th century'
    else:
        return '21st century'

df['century'] = df['year_published'].apply(get_century)
```
